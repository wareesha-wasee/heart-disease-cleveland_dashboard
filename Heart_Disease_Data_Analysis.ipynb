The box plot shows some potential outliers in the resting blood pressure column. These values are higher than the upper whisker. Further investigation is needed before deciding whether to remove them.

# IQR Method

```python
Q1 = df['trestbps'].quantile(0.25)
Q3 = df['trestbps'].quantile(0.75)

IQR = Q3 - Q1

lower_limit = Q1 - 1.5 * IQR
upper_limit = Q3 + 1.5 * IQR

outliers = df[(df['trestbps'] < lower_limit) | (df['trestbps'] > upper_limit)]

print("Q1:", Q1)
print("Q3:", Q3)
print("IQR:", IQR)
print("Lower Limit:", lower_limit)
print("Upper Limit:", upper_limit)
print("Total Outliers:", len(outliers))
```

Using the IQR method, 9 potential outliers were identified in the resting blood pressure column. The lower limit is 90 and the upper limit is 170.

```python
plt.figure(figsize=(8, 5))
sns.boxplot(x=df['chol'])
plt.title('Box Plot of Cholesterol')
plt.xlabel('Cholesterol Level')
plt.show()
```

```python
Q1 = df['chol'].quantile(0.25)
Q3 = df['chol'].quantile(0.75)

IQR = Q3 - Q1

lower_limit = Q1 - 1.5 * IQR
upper_limit = Q3 + 1.5 * IQR

outliers = df[(df['chol'] < lower_limit) | (df['chol'] > upper_limit)]

print("Q1:", Q1)
print("Q3:", Q3)
print("IQR:", IQR)
print("Lower Limit:", lower_limit)
print("Upper Limit:", upper_limit)
print("Total Outliers:", len(outliers))
```

The IQR method identified 5 potential outliers in the cholesterol column. The lower limit is 113.5 and the upper limit is 373.5.

# Outlier Handling

Potential outliers were identified in the trestbps and chol columns using the IQR method. These values were not removed because they may represent genuine patient measurements. Therefore, all outliers were retained to avoid losing important information.

# Univariate Analysis

```python
# Q1: What is the age distribution of patients?

plt.figure(figsize=(8, 5))
sns.histplot(df['age'], bins=15, kde=True)
plt.title('Age Distribution of Patients')
plt.xlabel('Age')
plt.ylabel('Number of Patients')
plt.show()
```

**Answer:**

 The age distribution shows that most patients are between approximately 50 and 65 years old. The highest frequency is observed around 57–60 years. This indicates that the dataset contains more middle-aged and older patients.

```python
 # Q2 What is the distribution of cholesterol levels?

plt.figure(figsize=(8, 5))
sns.histplot(df['chol'], bins=15, kde=True)
plt.title('Cholesterol Level Distribution')
plt.xlabel('Cholesterol Level')
plt.ylabel('Number of Patients')
plt.show()
```

**Answer:**

The histogram shows that most patients have cholesterol levels between approximately 200 and 300. The highest frequency is observed around 230–260. The distribution is right-skewed, with a few patients having very high cholesterol levels.

```python
# Q3: Which chest pain type is most common?

plt.figure(figsize=(8, 5))
sns.countplot(x=df['cp'])
plt.title('Distribution of Chest Pain Types')
plt.xlabel('Chest Pain Type')
plt.ylabel('Number of Patients')
plt.show()
```

**Answer:**

 The count plot shows that chest pain type 3 is the most common category, with approximately 142 patients. Chest pain type 0 is the least common category, with approximately 23 patients.

```python
 # Q4 How many patients belong to each condition category?

plt.figure(figsize=(8, 5))
sns.countplot(x=df['condition'])
plt.title('Distribution of Heart Disease Condition')
plt.xlabel('Condition Category')
plt.ylabel('Number of Patients')
plt.show()
```

**Answer:**

 The count plot shows that 160 patients belong to condition category 0, while 137 patients belong to condition category 1. Category 0 has a higher number of patients in this dataset.

# Bivariate Analysis

```python
# Q5 Is there any relationship between age and maximum heart rate?

plt.figure(figsize=(8, 5))
sns.scatterplot(x=df['age'], y=df['thalach'])
plt.title('Age vs Maximum Heart Rate')
plt.xlabel('Age')
plt.ylabel('Maximum Heart Rate')
plt.show()
```

```python
df['age'].corr(df['thalach'])
```

**Answer:**

 The scatter plot shows a negative relationship between age and maximum heart rate. The correlation value is approximately -0.395, indicating a weak-to-moderate negative correlation. Generally, as age increases, maximum heart rate tends to decrease.

```python
# Q6 Is there any relationship between cholesterol and resting blood pressure?

plt.figure(figsize=(8, 5))
sns.scatterplot(x=df['chol'], y=df['trestbps'])
plt.title('Cholesterol vs Resting Blood Pressure')
plt.xlabel('Cholesterol Level')
plt.ylabel('Resting Blood Pressure')
plt.show()
```

```python
df['chol'].corr(df['trestbps'])
```

**Answer:**

The scatter plot shows a very weak positive relationship between cholesterol and resting blood pressure. The correlation value is approximately 0.132, indicating that there is no strong linear relationship between these two variables.

# Correlation Heatmap

```python
plt.figure(figsize=(12, 8))

sns.heatmap(df.corr(), annot=True, cmap='coolwarm', fmt='.2f')

plt.title('Correlation Heatmap of Heart Disease Dataset')
plt.show()
```

The heatmap shows that condition has the strongest positive correlation with thal (0.52), followed by ca (0.46). It also has a negative correlation with thalach (-0.42). These values show relationships between the variables, but correlation does not prove causation.

```python
# Q7 How do average cholesterol levels compare across condition categories?

plt.figure(figsize=(8, 5))
sns.barplot(x='condition', y='chol', data=df)
plt.title('Average Cholesterol by Condition')
plt.xlabel('Condition Category')
plt.ylabel('Average Cholesterol Level')
plt.show()
```

```python
df.groupby('condition')['chol'].mean()
```

**Answer:**

 The average cholesterol level for condition category 0 is 243.49, while for condition category 1 it is 251.85. This shows that condition category 1 has a higher average cholesterol level than condition category 0.

```python
 # Q8 How does the condition category vary between male and female patients?
plt.figure(figsize=(8, 5))
sns.countplot(x='sex', hue='condition', data=df)
plt.title('Heart Disease Condition by Sex')
plt.xlabel('Sex')
plt.ylabel('Number of Patients')
plt.legend(title='Condition')
plt.show()
```

#**Answer**

 The count plot shows differences in condition categories across the two sex groups. In category 0, condition 0 has more patients, while in category 1, condition 1 has more patients. This indicates a difference in the distribution of condition categories between the two groups.
